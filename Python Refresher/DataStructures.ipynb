### Example 2 : Indexing and Slicing

````python
scores = [78, 85, 91, 65, 72, 88, 55]
````

Positive indexing - left to right, starts at 0

````python

 # 78  - first item
print(scores[0])
````

````python
# 91  - third item
print(scores[2])
````

````python
# 55  - last item (index = length - 1)
print(scores[6])
````

Negative indexing — right to left, starts at -1

````python
# 55  - last item
print(scores[-1])
````

````python
# 88  - second to last
print(scores[-2])
````

````python
# 78  : same as scores[0]
print(scores[-7])
````

Slicing: list[start : stop : step]  (stop is EXCLUSIVE)

````python
# [85, 91, 65]   — index 1, 2, 3
print(scores[1:4])
````

````python
# [78, 85, 91]   — first 3 items
print(scores[:3])
````

````python
# [65, 72, 88, 55] — from index 3 to end
print(scores[3:])
````

````python
# [78, 91, 72, 55] — every other item
print(scores[::2])
````

````python
# [55, 88, 72, 65, 91, 85, 78] — reversed!
print(scores[::-1])
````

Most Common List Error: IndexError
IndexError: list index out of range
This happens when you try to access an index that does not exist.

- scores = [78, 85, 91]   # only 3 items — valid indices: 0, 1, 2 (or -1, -2, -3)
- print(scores[3])        # IndexError! Index 3 does not exist.
- print(scores[5])        # IndexError! Way out of range.

Rule: valid positive indices are 0 to len(list)-1.

### Example 6 - List Comprehension
 	-concise way to create lists in python
 	-compact and easier to read than the traditional loops

````python
# The long way (using a for loop)
# Create a list of squares: [1,4,9,16,25]

squares = []
for x in range(1,6): # 1, 2, 3,4 ,5
  squares.append(x ** 2)
print(squares)
````

````python
# The short way (list comprehension)

# Pattern : [expression for item in iterable if condition]
#

squares = [x ** 2 for x in range(1,6)]
print(squares)
````

````python
# [expression  for  item  in  collection  if  condition]
scores = [78, 45, 92, 38, 65, 88, 55]
````

````python
# Get only passing scores (>= 50)
passing = [s for s in scores if s >= 50]
print("Passing:", passing)
````

### Example 8 : Tuple Unpacking

Tuple unpacking — assign multiple variables at once

````python
student = ("Amina Wanjiku", 24, "Data Science", 87.5)

name, age, track, gpa = student
print(f"Name: {name}")
print(f"Age: {age}")
print(f"Track: {track}")
print(f"GPA: {gpa}")
````

Unpacking in a loop : very common pattern

````python
students = [
    ("Amina",  87),
    ("Brian",  72),
    ("Njeri",  95),
]

for name, score in students:
    grade = "A" if score >= 80 else "B" if score >= 70 else "C"
    print(f"{name}: {score} → {grade}")
````

### Example 12 — List of Dictionaries (Real Data Pattern)

````python
# Most real data comes as a list of dictionaries — one dict per record
employees = [
    {"name": "Amina",  "dept": "Data",        "salary": 95000},
    {"name": "Brian",  "dept": "Engineering",  "salary": 85000},
    {"name": "Njeri",  "dept": "Data",         "salary": 102000},
    {"name": "Otieno", "dept": "Engineering",  "salary": 78000},
]
````

Print all employees

````python
for emp in employees:
    print(f"{emp['name']:<10} | {emp['dept']:<15} | Ksh {emp['salary']:,}")
````

# Filter — only Data department

````python
print("\nData team:")
data_team = [e for e in employees if e["dept"] == "Data"]
for emp in data_team:
    print(f"  {emp['name']}: Ksh {emp['salary']:,}")
````

Compute average salary

````python
salaries = [e["salary"] for e in employees]
avg = sum(salaries) / len(salaries)
print(f"\nAverage salary: Ksh {avg:,.0f}")
````

### Example 14 — Set Operations

Set operations — very useful for data analysis

````python
ds_students = {"Amina", "Brian", "Njeri", "Otieno"}
de_students = {"Njeri", "Otieno", "Kamau", "Aisha"}
````

Intersection — enrolled in BOTH tracks

````python
both = ds_students & de_students
print("Both tracks:", both)
````

Union — enrolled in EITHER track

````python
all_students = ds_students | de_students
print("All students:", all_students)
````

Difference —-only in DS (not DE)

````python
only_ds = ds_students - de_students
print("Only DS:", only_ds)
````

Symmetric difference - in one but not both

````python
exclusive = ds_students ^ de_students
print("Exclusive:", exclusive)
````

Real use case: find duplicate IDs in two datasets

````python
batch_1 = {101, 102, 103, 104, 105}
batch_2 = {103, 104, 105, 106, 107}
duplicates = batch_1 & batch_2
print("Duplicate IDs:", duplicates)
````

# Putting It All Together — Student Database

## This program uses lists, dictionaries, sets, loops, and functions together.

````python
students = [
    {"id": 1, "name": "Amina",  "track": "DS", "scores": [78, 85, 90]},
    {"id": 2, "name": "Brian",  "track": "DE", "scores": [65, 72, 68]},
    {"id": 3, "name": "Njeri",  "track": "DS", "scores": [91, 88, 95]},
    {"id": 4, "name": "Otieno", "track": "DE", "scores": [74, 79, 82]},
    {"id": 5, "name": "Wanjiku","track": "DS", "scores": [88, 92, 85]},
]

def get_average(scores):
    return round(sum(scores) / len(scores), 1)

def get_grade(avg):
    if avg >= 80: return "A"
    elif avg >= 70: return "B"
    elif avg >= 60: return "C"
    else: return "D"

# Print full report
print(f"{"Name":<10} {"Track":<6} {"Avg":>5} {"Grade":>6}")
print("-" * 30)

for s in students:
    avg   = get_average(s["scores"])
    grade = get_grade(avg)
    print(f"{s['name']:<10} {s['track']:<6} {avg:>5} {grade:>6}")

# Summary stats per track
tracks = set(s["track"] for s in students)
print()
for track in sorted(tracks):
    track_students = [s for s in students if s["track"] == track]
    avgs = [get_average(s["scores"]) for s in track_students]
    print(f"{track} track: {len(avgs)} students | Avg score: {round(sum(avgs)/len(avgs),1)}")
````
