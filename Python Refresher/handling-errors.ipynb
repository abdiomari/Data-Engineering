Q5.  Write validate_age(age) that raises TypeError if age is not an integer, and ValueError if age is not between 0 and 120. Wrap calls in try/except blocks to test both errors.

```python
num = ['4', '7', '', '2', 'N/A', '9', '0']
num_int = []
errors = []

for i, n in enumerate(num):
    try:
        number = int(n)
        num_int.append(number)
    except ValueError as e:
        errors.append((i, e))

print(num_int)
print(f"There are {len(num_int)} numbers in the list")
print(*errors, sep="\n")
```
