- Inheritance = Allows a class to inherit attributes and methods form another class   
- Helps with code reusability and extensibility   
- class Child(Parent)   

```python
class Animal: 
    def __init__(self, name) -> None:
        self.name = name
        self.isalive = True

    def eat(self):
        print(f"{self.name} is eating")
            
    def sleep(self):
        print(f"{self.name} is sleeping")

    def run(self):
        print(f"{self.name} is running")
```

```python
class Dog(Animal):
    pass

class Cat(Animal):
    pass

class Mouse(Animal):
    pass
```

```python
dog = Dog("Scooby")
cat = Cat("Garfield")
mouse = Mouse("Mickey")

print(dog.name)
dog.eat()
mouse.eat()
cat.eat()
```

```python
class Animal: 
    def __init__(self, name) -> None:
        self.name = name
        self.isalive = True

    def eat(self):
        print(f"{self.name} is eating")
            
    def sleep(self):
        print(f"{self.name} is sleeping")

    def run(self):
        print(f"{self.name} is running")


class Prey(Animal):
    def flee(self):
        print("This animal is fleeing")

class Predator:
    def hunt(self):
        print("This animal is hunting ")

class Rabbit(Prey):
    pass

class Hawk(Predator):
    pass

class Fish(Prey, Predator):
    pass
```

- Abstract class - class that cannot be instantiated on its own   
- they can contain abstract methods, which are declared but have no implementation
- Abstract classes benefits: 
    - Prevents instantiation of the class itself
    - Requires chiildren to use inherited abstract methods



```python
from abc import ABC, abstractmethod


class Vehicle(ABC):

    @abstractmethod
    def go(self):
        pass

    @abstractmethod
    def stop(self):
        pass
```

```python
class Car(Vehicle):
    def go(self):
        print("You can drive")
    
    def stop(self):
        print("You can stop")

car = Car()

car.go()
```

- super() - Function used in a child class to call methods from a parent class(super class)   
Allows you to extend the functionality of the inherited methods 

```python
class Shape:
    def __init__(self, color, filled) -> None:
        self.color = color
        self.filled = filled

    def describe(self):
        print(f"It is {self.color} and { 'filled' if self.filled else 'not filled'}")

class Circle(Shape):
    def __init__(self, color, filled, radius):
        super().__init__(color, filled)
        self.radius = radius

    def describe(self):
        super().describe()
        print(f"It is {self.color} circle with an area of {3.14 * self.radius * self.radius}")

class Square(Shape):
    def __init__(self, color, filled, width):
        super().__init__(color, filled)
        self.width = width

class Triangle(Shape):
    def __init__(self, color, filled, height, width):
        super().__init__(color, filled)
        self.height = height
        self.width = width


```

```python
circle = Circle("red", True, 5)
print(circle.color)

square = Square("blue", True, 6)
print(square.color)

triangle = Triangle("Green", False, 4, 8)
print(triangle.height)
```

- Polymorphism - Poly - Many and Morph - Form
- Inheritance and Duk typing

```python
from abc import ABC, abstractmethod

class Shape:

    @abstractmethod
    def area(self) -> float:
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side * self.side

class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height

```

```python
class Pizza(Circle):
    def __init__(self,topping, radius) -> None:
        super().__init__(radius)
        self.topping = topping
```

```python
shapes = [Circle(4), Square(4), Triangle(4,5), Pizza("mushrooms", 10)]

for shape in shapes:
    print(f"The area is {shape.area()} cm   ")
```

- Duck typing - Another way to achieve polymorphism besides Inheritance   
Objet must have the minimum necessary attributes/methods

" If it looks like a duck and quacks like a duck, it must be a duck"


```python
class Animal:
    alive = True

class Dog(Animal):
    def speak(self):
        print("Woof!")

class Cat(Animal):
    def speak(self):
        print("Woof!")

class Car:
    alive = True
    def speak(self):
        print("honk!")

```

```python
animals = [Dog(), Cat(), Car()]

for animal in animals:
    print(animal.speak())
```
