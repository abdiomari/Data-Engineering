## String methods
- strings in python come with built-in tools - called methods (transform or inspect string/text)

- String are immutable - methods do NOT change the original string. They return a NEW string.

- You call them using a dot: my_string.method()

```
name = "amina"
name.upper()
name = name.upper() -> store the result back
```

### Example 1 - case methods

````python
name = "aMina fatuma"

print(name.upper())     # ALL CAPS
print(name.lower())     # all lowercase
print(name.title())     # Title Case - first letter of each word capitalised
````

````python
city = input("Enter your city: ")
if city.lower() == "nairobi":
  print("Welcome to the capital")
````

### Example 2 - Stripping Whitespace
- the strip() method is used to remove characters from both the beginning and the end of a string. By default, it removes all leading and trailing whitespace, such as spaces, tabs (\t), and newlines (\n)

````python
raw = "   Nairobi  "

print(f"Original: [{raw}]")
print(f"strip(): [{raw.strip()}]")    # remove both sides
print(f"lstrip(): [{raw.lstrip()}]")  # left side only
print(f"rstrip(): [{raw.rstrip()}]")  # right side only
````

````python
# cleaning CSV data
record = "  Amina, 24, Nairobi   "
parts = [p.strip() for p in record.split(",")]
print(parts)
````

### Example 3 - replace(), find() , count()

- replace() method in Python is primarily used to swap parts of a string with new text.

```
string.replace(old, new, count)
```

- find() method in Python is used to locate the first occurrence of a specified substring within a string

```
string.find(substring, start, end)
```

- count() method is used to find the number of times a specific value appears in a collection, such as a list, string, or tuple



````python
sentence = "Python is fun. Python is p"
print(len(sentence))
````

````python
sentence = "Python is fun. Python is powerful. Python is everywhere"
print(sentence.find("powerful"))
````

````python
sentence1 = "Coding is fun. Coding is powerful. Coding is everywhere"
print(sentence1.find("powerful"))
````

````python
sentence = "Python is fun. Python is powerful. Python is everywhere"

# replace - swap one substring for another
print(sentence.replace("Python", "Coding"))

# find - returns index of first occurence (-1 if not found)
print(sentence.find("powerful"))
print(sentence.find("boring"))

# count - how many time does a substring appear?
print(sentence.count("Python"))
print(sentence.count("is"))
print(sentence.count("fun"))
````

### Example 4 - startswith() , endswith(), in

- The startswith() and endswith() methods in Python are used to check if a string begins or ends with a specific substring. Both methods return a Boolean (True or False)

````python
email = "amina.wanjiku@gmail.com"

print(email.startswith("amina"))
print(email.endswith(".com"))
print(email.endswith(".co.ke"))
````

````python
# in - check if substring is anywhere inside

print("gmail" in email)
print("yahoo" in email)
````

````python
# Real use: validate email domain

emails = ["brian@safaricom.co.ke", "njeri@gmail.com", "otieno@company.co.ke"]

for em in emails:
  if em.endswith(".co.ke"):
    print(f"  {em} - Kenyan domain")
  else:
    print(f"  {em} - International domain")
````

### Example 5 - split() and join()
- split() method in Python breaks a single string into a list of substrings based on a specified delimiter.

````python
# split() - break string into a list at every separator
csv_line = "Fatuma,24,Mombasa, Data Science"
fields = csv_line.split(",")
print(fields)
print("Name:", fields[0])
print("City:", fields[2])
````

````python
# split on space - default
sentence = "Python is great for data"
words = sentence.split()
print(words)
````

### join()
- join() method is a string method used to combine elements of an iterable (like a list, tuple, or set) into a single string, using a specific separator between each element


```
separator.join(iterable)
```

````python
# join() - opposite of split - glue a list back to a string

parts = ["Mombasa", "Nairobi", "Kisumu"]
result = " , ".join(parts)

print(result)
````

````python
# clean CSV line and re-join
# separator.join(iterable)
dirty = "   Amina  , 24 ,   Nairobi   "
clean = ",".join([d.strip() for d in dirty.split(",")])
print(clean)
````

## 2.  f-strings — Formatted String Literals

- starts with f before the quote
- anything inside {} is evaluated as Python - expressions, variable , method calls

- f-strings (formatted string literals) are a fast, readable way to insert variables and expressions directly into strings in Python

- use them instead of string concatenation (+) whenever possible

### Example f-string Basics and Number Formatting

````python
# Basic f-string
name   = "Amina"
age    = 24
city   = "Nairobi"
print(f"My name is {name}, I am {age} years old, from {city}.")
````

````python
# Expressions inside {}
a, b = 15, 4
print(f"{a} + {b} = {a + b}")
print(f"{a} divide by {b} = {a/b}")
````

````python
num = 4.2567
print(f"Num: {num:.1f}")
````

````python
# Number formatting
salary  = 85000
pi      = 3.14159265
big_num = 1234567

print(f"Salary: Ksh {salary:,}")
print(f"Pi: {pi:.2f}")
print(f"Big Number: {big_num:>12,}")
````

### Example f-string Alignment for Tables

````python
# {value:<width}  - left align
# {value:>width}  - right align
# {value:^width}  - centre align

employees = [
    ("Amina",   "Data Engineer",  95000),
    ("Brian",   "Developer",      85000),
    ("Njeri",   "Data Scientist", 102000),
    ("Otieno",  "Finance",        78000),
]

print(f"{"Name":<12} {"Role":<18} {"Salary":>8}")
print("-" * 42)

for name, role, salary in employees:
  print(f"{name:<12} {role:<18} Ksh {salary:>8,}")


````

````python
"""
▶ Output:
Name         Role               Salary
--------------------------------------------
Amina        Data Engineer      Ksh   95,000
Brian        Developer          Ksh   85,000
Njeri        Data Scientist     Ksh  102,000
Otieno       Finance            Ksh   78,000
"""
````

## 3.  File Handling — Reading and Writing Files

- 'r'mode - open to READ (you can look, not write)
- 'w'mode - open to WRITE (starts fresh - overwrite existing content)
- 'a' mode - open to APPEND (adds to the end - does not overwrite)
- 'r+' mode - open to READ and WRITE

ALWAYS use the with statement - it closes the file automatically, even if an error occurs

```
file_object = open("filename.txt", "mode")
```

### Example - writing to a file

- The write() method writes a specified text to the file.

- The writelines() method in Python is used to write a sequence of strings (like a list or tuple) to a file.

````python
# writing to a file - creates it if it does not exists

with open("students.txt", "w") as f:
  f.write("Amina Wanjiku\n")
  f.write("Navas Herbert\n")
  f.write("Nancy Mikia\n")
  f.write("Njeri Kamau\n")

print("File written successfully")
````

````python
names = ["Wanjiku Maina\n", "Otieno Achieng\n", "Moses Kibet\n"]

with open("more_students.txt", "w") as f:
  f.writelines(names)

print("Second file written")
````

the read() method is used to retrieve the contents of a file object. By default, it reads the entire file and returns it as a single string (in text mode)

````python
# read() - reads entire file as one string
with open("students.txt", "r") as f:
  content = f.read()
print(content)
````

The Python readlines() method reads all lines from a file and returns them as a list of strings, where each string is a single line including its trailing newline character (\n)


````python
# readlines() - read all lines into a list
with open("students.txt", "r") as f:
  lines = f.readlines()

print("--- readlines() ---")
print(lines)
````

````python
# loop over file directly - most memory effifient (best for big files)
print('--- Line by line ---')

with open("students.txt", "r") as f:
  for line in f:
    print(line.strip())   # strip() removes the \n at end of each line
````

````python
# Example - appending to a file

with open("students.txt", "a") as f:
  f.write("Moses Kibet\n")
````
